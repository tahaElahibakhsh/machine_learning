## 3) `for` Loops and `range`

A `for` loop is used to repeat an action over:
- numbers from `range(...)`
- a string
- a list (later sections)

Useful patterns:
- counting
- summing
- scanning characters

In this lesson, we will learn one of the most important tools in Python: the `for` loop.

A `for` loop is used when we want to **iterate** over a collection of items.

## What does "iterate" mean?

To iterate means:
- to visit items one by one
- to process elements in order
- to repeat an action for each element

For example:
- go through all numbers in a list
- go through all characters in a string
- repeat something several times using `range`

The `for` loop is extremely important in AI and data science because we often need to:
- process data samples
- count labels
- compute sums or averages
- examine words in text
- compare predictions with true answers

## What are we iterating over?

In a `for` loop, we always iterate over an **iterable**.

An **iterable** is an object that can produce its elements one by one.

Common iterables:
- lists
- strings
- tuples
- sets
- dictionaries
- `range(...)`

Examples:
- a list gives items
- a string gives characters
- a dictionary gives keys by default

## 8) Functions, First-Class Objects, `*args`, and `**kwargs`

Functions help us:
- avoid repetition
- organize code
- test smaller pieces
- write cleaner AI code

Important ideas:
- parameters
- return value
- default arguments
- functions are first-class objects
- flexible arguments with `*args` and `**kwargs`


```python
"""
Basic function examples.
"""

def square(x):
    return x * x

def greet(name="student"):
    return f"Hello, {name}!"

print(square(5))
print(greet())
print(greet("Arman"))
```

```python
"""
Returning multiple values.
"""

def min_max(values):
    return min(values), max(values)

mn, mx = min_max([4, 8, 1, 9, 2])
print(mn, mx)
```

```python
"""
Functions are first-class objects.
"""

def add_one(x):
    return x + 1

f = add_one
print(f(10))
```

```python
"""
Passing a function to another function.
"""

def apply_to_list(values, func):
    result = []
    for x in values:
        result.append(func(x))
    return result

def cube(x):
    return x ** 3

print(apply_to_list([1, 2, 3, 4], cube))
```

```python
"""
Using *args.
"""

def add_all(*args):
    total = 0
    for x in args:
        total += x
    return total

print(add_all(1, 2, 3))
print(add_all(10, 20, 30, 40))
```

```python
"""
Using **kwargs.
"""

def print_config(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} = {value}")

print_config(lr=0.01, batch_size=32, epochs=5)
```

## 9) Python Errors and Debugging

This part is essential. Good programmers do **not** avoid errors. They learn to:
1. read the error name
2. locate the line
3. understand what kind of mistake happened
4. fix the cause, not just the symptom

Common errors:
- `SyntaxError`
- `NameError`
- `TypeError`
- `ValueError`
- `IndexError`
- `KeyError`
- `ZeroDivisionError`

### What they mean
- `SyntaxError` → the code itself is not written in valid Python syntax
- `NameError` → you used a variable that does not exist
- `TypeError` → the operation does not fit the type
- `ValueError` → the type may be correct, but the content is not acceptable
- `IndexError` → list/string index out of range
- `KeyError` → dictionary key does not exist
- `ZeroDivisionError` → division by zero

`try/except` lets us handle some runtime errors safely.

## 11) Extra Essentials Before PyTorch

These are small topics, but they matter a lot later.

### `zip`
Useful when you want to iterate over two sequences together.

### Slicing review
Lists and strings both support slicing.

### Aliasing vs copying
`b = a` does **not** make a true copy for mutable objects like lists.

### `None`
Useful as a special value meaning “no value yet”.

### `deque`
Better than a list for real queue behavior.


```python
"""
zip example.
"""

preds = [1, 0, 1, 1]
labels = [1, 1, 1, 0]

correct = 0
for pred, label in zip(preds, labels):
    if pred == label:
        correct += 1

print(correct / len(preds))
```

```python
"""
Aliasing vs copying.
"""

a = [1, 2, 3]
b = a
b.append(4)

print("a =", a)
print("b =", b)

c = a[:]
c.append(5)

print("a =", a)
print("c =", c)
```

```python
"""
Using None safely in default-like behavior.
"""

def append_value(x, arr=None):
    if arr is None:
        arr = []
    arr.append(x)
    return arr

print(append_value(3))
print(append_value(7))
```

```python
"""
deque for queue behavior.
"""

from collections import deque

queue = deque()
queue.append("sample_1")
queue.append("sample_2")
queue.append("sample_3")

print(queue.popleft())
print(queue)
```
